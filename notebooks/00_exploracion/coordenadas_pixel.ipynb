```python
import lightkurve as lk
import matplotlib.pyplot as plt
import numpy as np

tic =  50897980

search = lk.search_tesscut(f"TIC {tic}")
tpf = search.download(cutout_size=18)#tamaño cuadrito

time = tpf.time.value
flux = tpf.flux

ny, nx = flux.shape[1:]

# WCS -> coordenadas del cubo
y_grid, x_grid = np.mgrid[:ny, :nx]
sky = tpf.wcs.pixel_to_world(x_grid, y_grid)

ra = sky.ra.deg
dec = sky.dec.deg

# mapa de brillo promedio
mean_flux = np.nanmean(flux, axis=0)

vmin = np.nanpercentile(mean_flux, 5)
vmax = np.nanpercentile(mean_flux, 95)

fig, axes = plt.subplots(ny, nx, figsize=(10,10))

for y in range(ny):
    for x in range(nx):

        ax = axes[ny-y-1, x]
        pixel_flux = flux[:, y, x]

        if np.all(np.isnan(pixel_flux)):
            ax.axis("off")
            continue

        # normalizar
        norm_flux = pixel_flux / np.nanmedian(pixel_flux)
        norm_flux = norm_flux - 1

        ax.plot(time, norm_flux, color="orange", linewidth=0.7)

        # fondo según brillo
        color = plt.cm.viridis((mean_flux[y,x] - vmin)/(vmax - vmin))
        ax.set_facecolor(color)

        ax.set_xticks([])
        ax.set_yticks([])

        # mostrar coordenadas en bordes
        if y == 0:  
            ax.set_xlabel(f"{ra[y,x]:.3f}", fontsize=6)

        if x == 0:
            ax.set_ylabel(f"{dec[y,x]:.3f}", fontsize=6)

plt.subplots_adjust(wspace=0, hspace=0)

fig.suptitle(f"TIC {tic} pixel light curves\nRA (bottom)  DEC (left)", fontsize=12)

plt.show()
```
