```python
import lightkurve as lk
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

search = lk.search_lightcurve("TIC 50897998", mission="TESS", author ="QLP")
print(search)
```

```python
lcs = search.download_all()
print(f"Curvas descargadas: {len(lcs)}")
```

```python
for lc in lcs:
    lc_clean = lc.remove_nans().normalize()
    fig = lc_clean.plot()
    fig.set_title(
        f"TIC {lc.targetid} – {lc.author} – Sector {lc.sector}"
    )
    plt.show()
```

```python
def plot_qlp_by_sector(tic_id, sectors=None):
    """
    Descarga curvas de luz QLP de TESS y grafica una por sector.

    Parámetros
    ----------
    tic_id : str o int
        Ejemplo: "TIC 50897998" o 50897998
    sectors : list o None
        Lista de sectores (ej. [6, 32, 98]).
        Si es None, descarga todos los sectores QLP disponibles.
    """

    search = lk.search_lightcurve(
        f"TIC {tic_id}",
        mission="TESS",
        author="QLP",
        sector=sectors
    )

    print(search)

    if len(search) == 0:
        print("⚠️ No se encontraron curvas QLP para ese TIC/sectores.")
        return

    lcs = search.download_all()
    print(f"Curvas descargadas: {len(lcs)}")

    for lc in lcs:
        lc_clean = lc.remove_nans().normalize()

        ax = lc_clean.plot()
        ax.set_title(
            f"TIC {lc.targetid} – {lc.author} – Sector {lc.sector}"
        )

        plt.show()
```

```python
plot_qlp_by_sector(138829403)
```
