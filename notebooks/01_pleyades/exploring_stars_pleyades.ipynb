```python

import pandas as pd

post1 = pd.read_csv("Tic_26017840_Posteriors.csv").drop(columns=["Unnamed: 0"])
post2 = pd.read_csv("TIC_35159657_Posteriors.csv").drop(columns=["Unnamed: 0"])
post3 = pd.read_csv("TIC_125754991_Posteriors.csv").drop(columns=["Unnamed: 0"])
post4 = pd.read_csv("TIC_346541355_Posteriors.csv").drop(columns=["Unnamed: 0"])
```

```python
print(post1.columns)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt
import numpy as np

fig, axs = plt.subplots(2,2, figsize=(10,6))

axs[0,0].hist(post1["B_porb"], bins=40)
axs[0,0].set_title("TIC 26017840")

axs[0,1].hist(post2["B_porb"], bins=40)
axs[0,1].set_title("TIC 35159657")

axs[1,0].hist(post3["B_porb"], bins=40)
axs[1,0].set_title("TIC 125754991")

axs[1,1].hist(post4["B_porb"], bins=40)
axs[1,1].set_title("Star 4")

plt.tight_layout()
plt.show()
```

```python
import numpy as np

summary = {
    "star": ["TIC 26017840", "TIC 35159657", "TIC 125754991", "TIC_346541355"],
    
    "P": [post1["B_porb"].mean(),
          post2["B_porb"].mean(),
          post3["B_porb"].mean(),
          post4["B_porb"].mean()],
    
    "R": [post1["B_r"].mean(),
          post2["B_r"].mean(),
          post3["B_r"].mean(),
          post4["B_r"].mean()],
    
    "M": [post1["B_m"].mean(),
          post2["B_m"].mean(),
          post3["B_m"].mean(),
          post4["B_m"].mean()],
    
    "ecc": [post1["B_ecc"].mean(),
            post2["B_ecc"].mean(),
            post3["B_ecc"].mean(),
            post4["B_ecc"].mean()]
}
```

```python
import pandas as pd

df = pd.DataFrame(summary)
#agregar los errores en esta tabla, desviacion estandar, cuartiles, investigar estadistica
# Redondear 
df = df.round(4)

print(df)
```

```python

plt.figure(figsize=(6,5))

plt.scatter(df["P"], df["R"])

for i in range(len(df)):
    plt.text(df["P"][i], df["R"][i], df["star"][i], fontsize=8)

plt.xlabel("Orbital Period (days)")
plt.ylabel("Secondary Radius")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.scatter(df["P"], df["ecc"])

for i in range(len(df)):
    plt.text(df["P"][i], df["ecc"][i], df["star"][i], fontsize=8)

plt.xlabel("Orbital Period (days)")
plt.ylabel("Eccentricity")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.scatter(df["P"], df["M"])

for i in range(len(df)):
    plt.text(df["P"][i], df["M"][i], df["star"][i], fontsize=8)

plt.xlabel("Orbital Period (days)")
plt.ylabel("Secondary Mass")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.scatter(df["M"], df["ecc"], s=80)

for i in range(len(df)):
    plt.annotate(df["star"][i], (df["M"][i], df["ecc"][i]), xytext=(5,5),
                 textcoords="offset points")

plt.xlabel("Secondary Mass ($M_\odot$)")
plt.ylabel("Eccentricity")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
exo = pd.read_csv("PS_2026.03.13_01.43.36.csv", comment="#")

exo = exo[[
    "pl_name",
    "pl_massj",
    "pl_orbper",
    "pl_orbeccen",
    "pl_radj"
]].dropna()

# convertir a masas solares
exo["mass_sun"] = exo["pl_massj"] / 1047.56
```

```python
my = pd.DataFrame({
"star":["TIC26017840","TIC35159657","TIC125754991","TIC346541355"],
"P":[1.385582,7.050250,2.461063,8.161548],
"M":[0.073640,0.073560,0.135428,1.500812],
"ecc":[0.018359,0.004413,0.000142,0.229899],
    "R": [0.141654, 0.171639, 0.539922, 0.881576]
})


plt.figure(figsize=(7,5))

plt.scatter(exo["mass_sun"],
            exo["pl_orbeccen"],
            s=10,
            alpha=0.3,
            label="Exoplanets")

plt.scatter(my["M"],
            my["ecc"],
            color="red",
            s=20,
            label="This work")

plt.xlabel("Mass (M☉)")
plt.ylabel("Eccentricity")
plt.xscale("log")

plt.legend()
plt.show()
```

```python
planets = exo[exo.pl_massj < 13]

bd = exo[(exo.pl_massj >= 13) & (exo.pl_massj < 78)]

stellar = exo[exo.pl_massj >= 78]



plt.figure(figsize=(7,5))

plt.scatter(planets.mass_sun,
            planets.pl_orbeccen,
            s=10,
            alpha=0.3,
            label="Planets")

plt.scatter(bd.mass_sun,
            bd.pl_orbeccen,
            s=20,
            alpha=0.6,
            label="Brown dwarfs")

plt.scatter(stellar.mass_sun,
            stellar.pl_orbeccen,
            s=20,
            alpha=0.3,
            label="Stellar companions")

plt.scatter(my.M,
            my.ecc,
            color="red",
            s=20,
            label="This work")

plt.axvline(13/1047.56, linestyle="--", color="gray")
plt.axvline(78/1047.56, linestyle="--", color="black")

plt.xscale("log")

plt.xlabel("Mass (M☉)")
plt.ylabel("Eccentricity")

plt.legend()
plt.show()
```
