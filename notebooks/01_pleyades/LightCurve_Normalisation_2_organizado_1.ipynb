```python
import numpy as np
import matplotlib.pylab as plt
import lightkurve as lk
from astropy.io import ascii

plt.rcParams.update({'font.size': 16})
```

```python
path = "/home/yceb2024/Lucia/starry/pleyades_2/TIC_26017840/"
file_o = "Tic 26017840_s_44_corrected.csv"
    
file_n = "tic_26017840_trend_wotan.csv"
```

```python
original = ascii.read(path+file_o, format='csv')

time_1 = original.columns[0].astype('float')
flux_1 = original.columns[2].astype('float')

trend = ascii.read(path+file_n, format='csv')

time_2 = trend.columns[0].astype('float')
flux_2 = trend.columns[1].astype('float')
```

```python
plt.figure(figsize=(10,4))
plt.scatter(time_1, flux_1, s=10, color='k')
plt.plot(time_2, flux_2, lw=2, color='r', ls='--')
plt.xlabel('Time MJD [days]')
plt.ylabel('Normalised Flux')
```

```python
curva = lk.LightCurve(time=time_1, flux=flux_1)
```

```python
curva_lc, curva_tr = curva.flatten(window_length=23, polyorder=2, sigma=3,
                                   break_tolerance=30, niters=100, return_trend=True)

#curva_lc = curva_lc[:-65]
```

```python
plt.figure(figsize=(10,4))
plt.scatter(time_1, flux_1, s=10, color='k')
plt.plot(curva_tr.time.value, curva_tr.flux.value, lw=2, color='r', ls='--')
plt.xlabel('Time MJD [days]')
plt.ylabel('Normalised Flux')
```

```python
plt.figure(figsize=(10,4))
plt.scatter(curva_lc.time.value, curva_lc.flux.value, s=5, color='b', ls='--')
plt.xlabel('Time MJD [days]')
plt.ylabel('Normalised Flux')
```
