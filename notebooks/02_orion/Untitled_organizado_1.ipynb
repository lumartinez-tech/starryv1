

The equilibrium point is given by $\rho = \rho_0$, where $P = 0$. At this point, the derivative is not zero but equal to $K_0 / \rho_0$, so it does not correspond to an extremum.

A local minimum occurs at $\rho = \rho_A$, obtained from $\frac{dP}{d\rho} = 0$, which gives:

$$
\frac{\rho_A}{\rho_0} = \left(\frac{w}{v}\right)^{\frac{1}{v - w}}
$$

This point lies between $0$ and $\rho_0$, and corresponds to the most negative pressure.

### General behavior

- For $\rho < \rho_0$: $P(\rho) < 0$ (attractive regime)
- At $\rho = \rho_A$: local minimum
- At $\rho = \rho_0$: $P = 0$
- For $\rho > \rho_0$: $P(\rho) > 0$ (repulsive regime)

The curve is asymmetric, with a steep increase for $\rho > \rho_0$ and a smoother decay for $\rho < \rho_0$.

## 6. Case $n = 4$ and $m = 2$

For this case:

$$
v = \frac{n+3}{3} = \frac{7}{3}, \quad w = \frac{m+3}{3} = \frac{5}{3}
$$

The EOS becomes:

$$
P(\rho) = \frac{3K_0}{2} \left[ \left(\frac{\rho}{\rho_0}\right)^{7/3} - \left(\frac{\rho}{\rho_0}\right)^{5/3} \right]
$$

---

### 

- The difference between exponents is smaller than in typical cases (e.g. $n=12, m=6$)
- The curve becomes **less steep**
- Repulsion grows more slowly at high densities
- The system is **more compressible**

---


```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
n, m = 4, 2
rho0 = 1.0
K0 = 1.0

# Exponentes
v = (n+3)/3
w = (m+3)/3

# Constante
u = 3*K0/(n-m)

# Rango de densidad
rho = np.linspace(0.01, 2.0, 500)

# EOS
P = u * ((rho/rho0)**v - (rho/rho0)**w)

# Punto mínimo
rho_A = rho0 * (w/v)**(1/(v-w))
P_A = u * ((rho_A/rho0)**v - (rho_A/rho0)**w)

# Gráfica
plt.figure()
plt.plot(rho, P, label="P(rho)")

# Línea de referencia
plt.axhline(0)

# Regiones sombreadas
plt.fill_between(rho, P, where=(rho < rho0), alpha=0.2, label="Expansion (rho < rho0)")
plt.fill_between(rho, P, where=(rho > rho0), alpha=0.2, label="Compression (rho > rho0)")

# Punto mínimo
plt.scatter(rho_A, P_A, color='red', zorder=3, label='Minimum')

# Línea en rho0
plt.axvline(rho0, linestyle='--', label='rho0')

plt.xlabel("rho / rho0")
plt.ylabel("P(rho)")
plt.title("EOS P(rho) with regimes")
plt.legend()

plt.show()

print("rho_A =", rho_A)
```
