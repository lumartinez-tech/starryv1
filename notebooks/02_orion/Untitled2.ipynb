```python
import read_mist_models

iso = read_mist_models.ISO('home/luciam/Documents/catalogos_tesis/mist_isochrones/MIST_iso_6a9a3d9369996.iso')

print(iso.ages)          # lista de log10(edad) disponibles
print(iso.hdr_list)      # nombres de columnas

# Opción A: exportar UNA edad a CSV para TOPCAT
import pandas as pd
age_index = iso.age_index(9.0)  # ejemplo: log(edad)=9.0
df = pd.DataFrame(iso.isos[age_index])
df.to_csv('isocrona_log_age_9.0.csv', index=False)

# Opción B: exportar TODAS las edades en una sola tabla plana (con columna log_age)
frames = []
for i, age in enumerate(iso.ages):
    d = pd.DataFrame(iso.isos[i])
    d['log_age'] = age
    frames.append(d)
todas = pd.concat(frames, ignore_index=True)
todas.to_csv('todas_las_isocronas.csv', index=False)
```
