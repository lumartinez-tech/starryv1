```python
import lightkurve as lk
import matplotlib.pyplot as plt

search = lk.search_lightcurve("TIC 11300044",
                              mission="TESS",
                              author="QLP")

print(search)

for i in range(len(search)):
    lc = search[i].download(cache=False)  # 👈 clave
    
    print(f"Sector {search[i].mission} descargado")
    
    lc.plot()
    plt.show()
    
    del lc  
```
